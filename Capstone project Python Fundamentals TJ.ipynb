```python
# Task 2 

# Read the Project DataFrame from the CSV file
project_df = pd.read_csv("Project_DataFrame.csv")
```

```python
# Display the Project DataFrame
project_df
```

```python
# Initialize an empty list to collect non-missing cost values
cost_list = []

# Loop through each row of the Project DataFrame
for i in range(len(project_df)):

    # If cost is not missing
    if pd.notna(project_df.loc[i, "Cost"]):
        # Add the current Cost to the list
        cost_list.append(project_df.loc[i, "Cost"])

    # If cost is missing
    else:
        # Compute running average using numpy so far
        running_avg = np.mean(cost_list)
        # Replace the missing value with the running average
        project_df.loc[i, "Cost"] = running_avg
        # Add this computed value to the list too, so it's included in the next running average
        cost_list.append(running_avg)
```

```python
# Display the updated Project DataFrame after replacing missing values
project_df
```

```python
# Task 3 

# Read the Employee DataFrame from the CSV file
employee_df = pd.read_csv("Employee_DataFrame.csv")
```

```python
# Display the Employee DataFrame
employee_df
```

```python
# Split the Name column into First Name and Last Name columns
employee_df[["First Name", "Last Name"]] = employee_df["Name"].str.split(" ", n=1, expand=True)
```

```python
# Remove the original Name column from the DataFrame
employee_df = employee_df.drop("Name", axis=1)
```

```python
# Display the updated Employee DataFrame
employee_df
```

```python
# Task 4 

# Read the Seniority Level DataFrame from the CSV file
seniority_df = pd.read_csv("Seniority_DataFrame.csv")
```

```python
# Merge the Project and Employee DataFrames using the ID column
Final = pd.merge(project_df, employee_df, on="ID")
```

```python
# Merge the Final DataFrame with the Seniority DataFrame using the ID column
Final = pd.merge(Final, seniority_df, on="ID")
```

```python
# Display the merged Final DataFrame
Final 
```

```python
# Task 5
# Calculate a 5% bonus for employees whose project status is Finished, using np.where
Final["Bonus"] = np.where(Final["Status"] == "Finished", Final["Cost"] * 0.05, 0)
```

```python
# Display the Final DataFrame with the Bonus column
Final 
```

```python
# Task 6

# find the set of employee IDs that have at least one Failed project
failed_ids = Final.loc[Final["Status"] == "Failed", "ID"].unique()
```

```python
# demote by 1 — apply to EVERY row of an employee who has any failed project, so the same ID gets a consistent Designation Level across all its rows
Final["Designation Level"] = np.where(
    Final["ID"].isin(failed_ids),
    Final["Designation Level"] + 1,
    Final["Designation Level"]
)
```

```python
# find IDs (not just rows) whose Designation Level is now above 4
ids_to_drop = Final.loc[Final["Designation Level"] > 4, "ID"].unique()
```

```python
# drop the entire employee (all their rows), not just the offending row
Final = Final[~Final["ID"].isin(ids_to_drop)]
```

```python
# Display the updated Final DataFrame after modifying the Designation Level
Final
```

```python
# Task 7 

# Add the prefix "Mr." to the First Name of all male employees
Final.loc[Final["Gender"] == "M", "First Name"] = "Mr. " + Final["First Name"]
```

```python
# Add the prefix "Mrs." to the First Name of all female employees
Final.loc[Final["Gender"] == "F", "First Name"] = "Mrs. " + Final[ "First Name"]
```

```python
# Remove the Gender column from the Final DataFrame
Final = Final.drop("Gender", axis=1)
```

```python
# Display the updated Final DataFrame
Final 
```

```python
# Task 8

# Promote designation level by 1 for the employees whose age is more than 29 years using IF condition 
def promote(row):
    # If employee's age is more than 29, promote (lower number = higher rank on this scale)
    if row["Age"] > 29:
        return max(row["Designation Level"] - 1, 1)  # promote, but never go below level 1
    else:
        return row["Designation Level"]
```

```python
# Apply the function row-wise using the IF condition above
Final["Designation Level"] = Final.apply(promote, axis=1)
```

```python
# Display the updated Final DataFrame after promoting eligible employees
Final
```

```python
# Task 9 

# Calculate the total project cost for each employee by grouping the data using ID and First Name
TotalProjCost = Final.groupby(["ID", "First Name"])["Cost"].sum().reset_index()
```

```python
# Rename the Cost column to Total Cost
TotalProjCost.rename(columns={"Cost": "Total Cost"}, inplace=True) 
```

```python
# Display the Total Project Cost DataFrame
TotalProjCost
```
